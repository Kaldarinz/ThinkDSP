```python
import numpy as np
import thinkdsp
from matplotlib import pyplot as plt
```

```python
saxo = thinkdsp.read_wave('100475__iluppai__saxophone-weep.wav')
saxo.plot()
```

```python
saxo.make_spectrum().plot()
```

```python
sig = np.zeros_like(saxo.ys)
sig[::4] = 1
train = thinkdsp.Wave(ys=sig, framerate=saxo.framerate)
train.make_spectrum(full=True).plot()
```

```python
(saxo*train).make_spectrum().plot()
```

```python
saxo.make_spectrum(full=True).convolve(train.make_spectrum(full=True)).plot()
```

```python
wave1 = thinkdsp.SquareSignal().make_wave()
wave2 = thinkdsp.SquareSignal().make_wave()
wave1.segment(duration=.01).plot()
```

```python
conv = np.convolve(wave1.ys, wave2.ys, mode='same')
plt.plot(conv)
plt.gca().set_xlim()
```

```python
mult = np.fft.irfft(wave1.make_spectrum().hs**2)
plt.plot(mult)
plt.gca().set_xlim()
```

```python
tot_dur = 1
sig_dur = .05
points = 2048

wave = np.zeros(points)
wave[:int(points*sig_dur)] = 1
plt.plot(wave)
```

```python
plt.plot(np.abs(np.fft.rfft(wave)))
```

```python
shifted_wave = np.roll(wave, 500)
plt.plot(shifted_wave)
```

```python
plt.plot(np.abs(np.fft.rfft(shifted_wave)))
```

```python
cut_wave = wave[:points//4]
plt.plot(cut_wave)
```

```python
plt.plot(np.abs(np.fft.rfft(cut_wave)))
```
